```python
import os
import urllib.request

# Direct GitHub URLs for UNSW-NB15 training and testing sets
# Source: https://github.com/sarax26/UNSW-NB15-NIDS
train_url = "https://raw.githubusercontent.com/sarax26/UNSW-NB15-NIDS/main/Dataset/UNSW_NB15_training-set.csv"
test_url  = "https://raw.githubusercontent.com/sarax26/UNSW-NB15-NIDS/main/Dataset/UNSW_NB15_testing-set.csv"

if not os.path.exists("UNSW_NB15_training-set.csv"):
    urllib.request.urlretrieve(train_url, "UNSW_NB15_training-set.csv")
    print("✅ UNSW-NB15 training set downloaded.")
else:
    print("ℹ️ UNSW-NB15 training set already exists.")

if not os.path.exists("UNSW_NB15_testing-set.csv"):
    urllib.request.urlretrieve(test_url, "UNSW_NB15_testing-set.csv")
    print("✅ UNSW-NB15 testing set downloaded.")
else:
    print("ℹ️ UNSW-NB15 testing set already exists.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the training and testing sets (downloaded in Cell 8 via GitHub)
train_unsw = pd.read_csv("UNSW_NB15_training-set.csv")
test_unsw  = pd.read_csv("UNSW_NB15_testing-set.csv")

# The GitHub CSVs sometimes include an unnamed index column at the start — drop it if present
for df in [train_unsw, test_unsw]:
    if df.columns[0].startswith("Unnamed") or df.columns[0] == "":
        df.drop(df.columns[0], axis=1, inplace=True)

print(f"✅ UNSW-NB15 Training set shape: {train_unsw.shape}")
print(f"✅ UNSW-NB15 Test set shape: {test_unsw.shape}")
print("\nColumns:", list(train_unsw.columns))
print("\nFirst 3 rows:")
display(train_unsw.head(3))

# Class distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
train_unsw['label'].value_counts().plot(kind='bar', ax=axes[0], color=['#2ecc71', '#e74c3c'])
axes[0].set_title('Training Set – Label Distribution')
axes[0].set_xticklabels(['Normal', 'Attack'], rotation=0)

train_unsw['attack_cat'].value_counts().plot(kind='bar', ax=axes[1], color='steelblue')
axes[1].set_title('Training Set – Attack Categories')
plt.tight_layout()
plt.show()
```

```python
# Drop columns that are not useful for prediction
# 'id' is just a row index; 'attack_cat' is the multi-class label (we use binary 'label' here)
drop_cols = ['id', 'attack_cat']
train_unsw = train_unsw.drop(columns=[c for c in drop_cols if c in train_unsw.columns])
test_unsw  = test_unsw.drop(columns=[c for c in drop_cols if c in test_unsw.columns])

# Separate features and binary target
X_train_u = train_unsw.drop('label', axis=1)
y_train_u = train_unsw['label']
X_test_u  = test_unsw.drop('label', axis=1)
y_test_u  = test_unsw['label']

# Identify categorical columns (only these three exist in UNSW-NB15)
cat_cols_u = ['proto', 'service', 'state']

# One-hot encode categorical features
X_train_u = pd.get_dummies(X_train_u, columns=cat_cols_u)
X_test_u  = pd.get_dummies(X_test_u,  columns=cat_cols_u)

# Align columns: ensure both sets have the same dummy columns (fixes unseen categories)
X_train_u, X_test_u = X_train_u.align(X_test_u, join='left', axis=1, fill_value=0)

# Handle any remaining NaN values (safety net)
X_train_u = X_train_u.fillna(0)
X_test_u  = X_test_u.fillna(0)

# Scale numerical features
from sklearn.preprocessing import StandardScaler
scaler_u = StandardScaler()
X_train_u_scaled = scaler_u.fit_transform(X_train_u)
X_test_u_scaled  = scaler_u.transform(X_test_u)

print("✅ UNSW-NB15 preprocessing complete.")
print(f"Training features shape: {X_train_u_scaled.shape}")
print(f"Test features shape: {X_test_u_scaled.shape}")
```

```python
from sklearn.ensemble import RandomForestClassifier

# Train a Random Forest classifier (same architecture as NSL-KDD)
rf_unsw = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    random_state=42,
    n_jobs=-1
)

rf_unsw.fit(X_train_u_scaled, y_train_u)
print("✅ Random Forest model trained on UNSW-NB15.")
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Predict on the UNSW-NB15 test set
y_pred_u = rf_unsw.predict(X_test_u_scaled)

# Accuracy
acc_u = accuracy_score(y_test_u, y_pred_u)
print(f"🎯 UNSW-NB15 Test Accuracy: {acc_u:.4f}\n")

# Classification report
print("📊 Classification Report (UNSW-NB15):")
print(classification_report(y_test_u, y_pred_u, target_names=['Normal', 'Attack']))

# Confusion matrix
cm_u = confusion_matrix(y_test_u, y_pred_u)
plt.figure(figsize=(6, 4))
sns.heatmap(cm_u, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Normal', 'Attack'],
            yticklabels=['Normal', 'Attack'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix – UNSW-NB15 Test Set')
plt.show()
```

```python
# Top 15 important features from UNSW-NB15
feat_imp_u = pd.Series(
    rf_unsw.feature_importances_,
    index=X_train_u.columns
).sort_values(ascending=False).head(15)

plt.figure(figsize=(10, 6))
sns.barplot(x=feat_imp_u.values, y=feat_imp_u.index, palette='magma')
plt.xlabel('Importance Score')
plt.title('Top 15 Important Features – UNSW-NB15')
plt.tight_layout()
plt.show()
```

```python
# Compare accuracy of both models
comparison = pd.DataFrame({
    'Dataset': ['NSL-KDD', 'UNSW-NB15'],
    'Accuracy': [accuracy, acc_u],
    'Training Samples': [X_train_scaled.shape[0], X_train_u_scaled.shape[0]],
    'Test Samples': [X_test_scaled.shape[0], X_test_u_scaled.shape[0]],
    'Features': [X_train_scaled.shape[1], X_train_u_scaled.shape[1]]
})

print("📊 Model Comparison:")
display(comparison)

# Bar plot comparison
plt.figure(figsize=(8, 4))
sns.barplot(x='Dataset', y='Accuracy', data=comparison, palette='coolwarm')
plt.ylim(0.8, 1.0)
plt.title('Accuracy Comparison: NSL-KDD vs UNSW-NB15')
plt.ylabel('Accuracy')
for i, v in enumerate(comparison['Accuracy']):
    plt.text(i, v + 0.005, f"{v:.4f}", ha='center', fontweight='bold')
plt.show()
```

```python
# Reload the raw CSVs to keep 'attack_cat' for multi-class classification
train_multi = pd.read_csv("UNSW_NB15_training-set.csv")
test_multi  = pd.read_csv("UNSW_NB15_testing-set.csv")

# Drop unnamed index column if present
for df in [train_multi, test_multi]:
    if df.columns[0].startswith("Unnamed") or df.columns[0] == "":
        df.drop(df.columns[0], axis=1, inplace=True)

# Keep only attack categories (drop normal traffic for multi-class)
train_multi = train_multi[train_multi['attack_cat'] != 'Normal'].copy()
test_multi  = test_multi[test_multi['attack_cat'] != 'Normal'].copy()

# Strip whitespace from attack_cat (some versions have trailing spaces)
train_multi['attack_cat'] = train_multi['attack_cat'].str.strip()
test_multi['attack_cat']  = test_multi['attack_cat'].str.strip()

# Prepare features
X_train_m = train_multi.drop(['id', 'label', 'attack_cat'], axis=1, errors='ignore')
y_train_m = train_multi['attack_cat']
X_test_m  = test_multi.drop(['id', 'label', 'attack_cat'], axis=1, errors='ignore')
y_test_m  = test_multi['attack_cat']

# One-hot encode categorical features
X_train_m = pd.get_dummies(X_train_m, columns=['proto', 'service', 'state'])
X_test_m  = pd.get_dummies(X_test_m,  columns=['proto', 'service', 'state'])
X_train_m, X_test_m = X_train_m.align(X_test_m, join='left', axis=1, fill_value=0)

# Fill NaNs (safety)
X_train_m = X_train_m.fillna(0)
X_test_m  = X_test_m.fillna(0)

# Scale
scaler_m = StandardScaler()
X_train_m_scaled = scaler_m.fit_transform(X_train_m)
X_test_m_scaled  = scaler_m.transform(X_test_m)

# Train a multi-class Random Forest
rf_multi = RandomForestClassifier(n_estimators=100, max_depth=20,
                                   random_state=42, n_jobs=-1)
rf_multi.fit(X_train_m_scaled, y_train_m)

# Evaluate
y_pred_m = rf_multi.predict(X_test_m_scaled)
print(f"🎯 Multi-Class Attack Classification Accuracy: {accuracy_score(y_test_m, y_pred_m):.4f}\n")
print("📊 Classification Report (Attack Categories):")
print(classification_report(y_test_m, y_pred_m))
```

```python
# Visualize which attack categories the model predicts vs. actual
from sklearn.metrics import confusion_matrix

labels = sorted(y_test_m.unique())
cm_multi = confusion_matrix(y_test_m, y_pred_m, labels=labels)

plt.figure(figsize=(10, 7))
sns.heatmap(cm_multi, annot=True, fmt='d', cmap='YlOrRd',
            xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted Attack Category')
plt.ylabel('Actual Attack Category')
plt.title('Multi-Class Attack Classification – Confusion Matrix')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
